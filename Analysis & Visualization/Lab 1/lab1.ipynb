## Task 1

```python
central = ["Södermalms", "Östermalms", "Norrmalms", "Kungsholmens"]

mask = listings['neighbourhood_cleansed'].isin(central)
```

```python
listings.shape
```

How many central listings there are. (Should give 2331 rows.)

```python
#Mask 
print(mask.sum())
print(mask.mean().round(3))
```

How many outside listings there are. (Should give 2243 rows.)

```python
#Inverted mask
listings[~mask].shape
```

#### Private rooms: Central vs. Outside

How many central listings there are that are private. (Should give sum: 352, share: 0.151.)

```python
#Central
print((listings[mask]['room_type'] == "Private room").sum())
print((listings[mask]['room_type'] == "Private room").mean().round(3))
```

How many outside listings there are that are private. (Should give sum: 435, share: 0.194.)

```python
#Outside
print((listings[~mask]['room_type'] == "Private room").sum())
print((listings[~mask]['room_type'] == "Private room").mean().round(3))
```

Outside listings have a bigger proportion of private listings.

## Task 2

Hint for myself: About 0.557 of the entire homes and about 0.093 of the other listings.

Share of Entire home/apt listings with room for four guests or more (accommodates). (Should give 0.557)

```python
mask = listings['room_type'] == "Entire home/apt"

share = (listings[mask]['accommodates'] >= 4).mean()

print(share.round(3))
```

Share of listings that are not Entire home/apt with room for four guests or more (accommodates). (Should give 0.093)

```python
mask = listings['room_type'] != "Entire home/apt"

share = (listings[mask]['accomodates'] >= 4).mean()

print(share.round(3))
```
