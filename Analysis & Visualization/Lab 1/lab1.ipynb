```python
import pandas as pd
listings = pd.read_csv('listings.csv')
```

## Task 1

```python
central = ["Södermalms", "Östermalms", "Norrmalms", "Kungsholmens"]

mask = listings['neighbourhood_cleansed'].isin(central)
```

```python
listings.shape
```

How many central listings there are. (Should give 2331 rows.)

```python
#Mask 
print(mask.sum())
print(mask.mean().round(3))
```

```python
How many outside listings there are. (Should give 2243 rows.)
```

```python
#Inverted mask
listings[~mask].shape
```

#### Private rooms: Central vs. Outside

```python
#Central
print((listings[mask]['room_type'] == "Private room").sum())
print((listings[mask]['room_type'] == "Private room").mean().round(3))
```

```python
#Outside
print((listings[~mask]['room_type'] == "Private room").sum())
print((listings[~mask]['room_type'] == "Private room").mean().round(3))
```
