## Task 2

Hint for myself: About 0.557 of the entire homes and about 0.093 of the other listings.

Share of Entire home/apt listings with room for four guests or more (accommodates). (Should give 0.557)

```python
mask = listings['room_type'] == "Entire home/apt"

share = (listings[mask]['accommodates'] >= 4).mean()

print(share.round(3))
```

Share of listings that are not Entire home/apt with room for four guests or more (accommodates). (Should give 0.093)

```python
mask = listings['room_type'] != "Entire home/apt"

share = (listings[mask]['accommodates'] >= 4).mean()

print(share.round(3))
```

The difference shows that room types that are not of as great size rarely allow accommodation for many people.
Which is realistic; 4 or more people on a smaller space means that chances are that the shared area becomes more cramped.
